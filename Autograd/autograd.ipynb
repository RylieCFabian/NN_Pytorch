```python
import torch

x = torch.ones(5)
y = torch.zeros(3)
w = torch.randn(5 , 3 , requires_grad= True)
b = torch.randn(3, requires_grad= True)
z = torch.matmul(x , w) + b
loss = torch.nn.functional.binary_cross_entropy_with_logits(z , y)

print(f"Gradiant Function for z = {z.grad_fn}")
print(f"Gradiant Function for loss = {loss.grad_fn}")

loss.backward()
print(w.grad)
print(b.grad)
```

```python
z = torch.matmul(x , w) + b
print(z.requires_grad)

with torch.no_grad():
    z = torch.matmul(x , w) + b
print(z.requires_grad)
```
